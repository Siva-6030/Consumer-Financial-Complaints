### Missing Value Analysis

```python
missing_values = clean_data.isnull().sum()

missing_percentage = (
    clean_data.isnull().sum() / len(clean_data) * 100
).round(2)

missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": missing_percentage
})
missing_values
```

### Visualize Missing Values

```python
missing_plot = missing_summary[
    missing_summary["Missing Values"] > 0
]

plt.figure(figsize=(12, 6))

plt.bar(
    missing_plot.index,
    missing_plot["Missing Percentage"]
)

plt.xlabel("Columns")
plt.ylabel("Missing Values (%)")
plt.title("Missing Values by Column")

plt.xticks(rotation=75)

plt.tight_layout()
plt.show()
```

### Standardize Column Names

```python
clean_data.columns = (
    clean_data.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("-", "_")
    .str.replace("?", "", regex=False)
)

print("Updated Column Names:\n")

for i, column in enumerate(clean_data.columns, start=1):
    print(f"{i} -> {column}")
```

### Check Data Types

```python
print(clean_data.dtypes)
```

### Convert Date Columns

```python
clean_data["date_received"] = pd.to_datetime(
    clean_data["date_received"],
    errors="coerce"
)

clean_data["date_sent_to_company"] = pd.to_datetime(
    clean_data["date_sent_to_company"],
    errors="coerce"
)

clean_data["zip_code"] = pd.to_numeric(
    clean_data["zip_code"],
    errors="coerce"
)

print(clean_data[
    ["date_received", "date_sent_to_company","zip_code"]
].dtypes)
```

### Clean Text Columns

```python
text_columns = [
    "product",
    "sub_product",
    "issue",
    "sub_issue",
    "company_public_response",
    "company",
    "state",
    "tags",
    "submitted_via",
    "company_response_to_consumer",
    "timely_response"
]

for column in text_columns:
    clean_data[column] = (
        clean_data[column]
        .astype("string")
        .str.strip()
    )

print("Text columns cleaned successfully.")
```

### Standardize State Values

```python
clean_data["state"] = (
    clean_data["state"]
    .str.upper()
    .str.strip()
)

clean_data["state"].value_counts().head(10)
```

### Check Complaint IDs

```python
print("Total Complaint IDs:",
      len(clean_data))

print("Unique Complaint IDs:",
      clean_data["complaint_id"].nunique())

print(
    "Duplicate Complaint IDs:",
    clean_data["complaint_id"].duplicated().sum()
)
```

### Create Response Time

```python
clean_data["response_days"] = (clean_data["date_sent_to_company"]- clean_data["date_received"]).dt.days

clean_data[["date_received","date_sent_to_company","response_days"]].head()
```

### Create Response Time

```python
negative_response = (
    clean_data["response_days"] < 0
).sum()

print(
    "Negative response-time records:",
    negative_response
)
```

```python
product_response_time = (
    clean_data
    .groupby("product")["response_days"]
    .agg([
        "count",
        "mean",
        "median"
    ])
    .sort_values(
        "count",
        ascending=False
    )
)

product_response_time.head(15)
```

```python
top_products = (
    clean_data["product"]
    .value_counts()
    .head(10)
    .index
)

plot_data = (
    product_response_time
    .loc[top_products]
    .sort_values("mean")
)

plt.figure(figsize=(12, 7))

plt.barh(
    plot_data.index,
    plot_data["mean"]
)

plt.xlabel("Average Response Time (Days)")
plt.ylabel("Product")
plt.title("Average Company Response Time by Product")

plt.tight_layout()
plt.show()
```
